```python
import pandas as pd
from pathlib import Path
df = pd.read_csv('../data/processed/cleaned_sales.csv')
df.head()
```

```python
df.info()
df[['Gross_Sales','Discount_Amount','Net_Sales']].describe()
```

```python
df.groupby('Branch', as_index=False)['Net_Sales'].sum().sort_values('Net_Sales', ascending=False)
```

```python
df.groupby('Month', as_index=False)['Net_Sales'].sum().plot(x='Month', y='Net_Sales', marker='o', title='Monthly Net Sales')
```
